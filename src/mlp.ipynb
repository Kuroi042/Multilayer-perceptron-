```python
features = [
    'radius_mean',
    'perimeter_mean',
    'area_mean',
    'concavity_mean',
    'concave_points_mean'
]

for feature in features:
    plt.figure(figsize=(8, 5))

    sns.boxplot(data=df, x='diag', y=feature)

    plt.title(f'{feature}: Benign vs Malignant')
    plt.xlabel('Diagnosis')
    plt.ylabel(feature)

    plt.show()
```

```python

features = df.drop(columns=['id', 'diag'])

# Calculate Pearson correlation
correlation_matrix = features.corr()

# Create the heatmap
plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    annot=True,
    fmt='.1f',
    linewidths=0.5
)

plt.title('Correlation Matrix of WDBC Features')
plt.tight_layout()
plt.show()
```

```python
# Convert diagnosis into numerical values
df['diag_encoded'] = df['diag'].map({'B': 0, 'M': 1})

# Select the 30 features and the encoded diagnosis
features = df.drop(columns=['id', 'diag', 'diag_encoded'])

# Calculate correlation with diagnosis
diagnosis_corr = features.corrwith(df['diag_encoded'])

# Sort by absolute correlation, highest first
diagnosis_corr = diagnosis_corr.reindex(
    diagnosis_corr.abs().sort_values(ascending=False).index
)

print(diagnosis_corr)
df
```
